```python
import screens as sc

def main():
    sc.show_developer_info()
    input("\nplease press any key to continue...")
    print(f"\nWelcome to the SLMS")

    while True:
        # display the menu and get user choice
        sc.display_main_menu()

        # get user choice and call the appropriate screen function
        choice = sc.get_user_choice()

        # call the appropriate screen function based on the user's choice
        match choice:
            case 'A':
                sc.add_student()
            case 'B':
                sc.modify_student_info()
            case 'C':
                sc.show_reports_screen()
            case 'D':
                print("\nExiting the program. Goodbye!")
                break
```

```python
main()
```
